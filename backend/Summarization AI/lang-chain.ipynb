```python
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate

# 1. Load model (OpenAI GPT)
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# 2. Create a prompt
prompt = ChatPromptTemplate.from_template("Translate this sentence to French, Hindi and Arabic, Show in bullet form: {sentence}")

# 3. Build chain
chain = prompt | llm

# 4. Run it
result = chain.invoke({"sentence": "Hello, how are you?"})
print(result.content)
```

```python
from langchain_openai import ChatOpenAI
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# Add memory
memory = ConversationBufferMemory()

# Create chain with memory
conversation = ConversationChain(llm=llm, memory=memory, verbose=True)

print(conversation.run("Hello, I’m learning LangChain."))
print(conversation.run("Can you remind me what I just said?"))
```
