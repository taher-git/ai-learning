```python
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate
from langchain.chains import LLMChain

# Prompt for summarization
prompt = ChatPromptTemplate.from_template("""
You are a meeting summarizer.
Summarize the following transcript into {style}:

Transcript:
{transcript}
""")
# Load LLM
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# Create summarization chain
summarize_chain = LLMChain(llm=llm, prompt=prompt)

transcript = """
Today we discussed project deadlines, budget allocation, 
and assigning tasks to the frontend and backend teams. 
Next steps include preparing the design mockups and scheduling the next sprint planning.
"""

# Example 1: Paragraph summary
result_paragraph = summarize_chain.invoke({"transcript": transcript, "style": "a short paragraph"})
print("\nParagraph Summary:\n", result_paragraph["text"])

# Example 2: Bullet summary
result_bullet = summarize_chain.invoke({"transcript": transcript, "style": "bullet points"})
print("\nBullet Summary:\n", result_bullet["text"])
```
