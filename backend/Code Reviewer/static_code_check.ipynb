```python
import re

def static_checks(java_code):
    """
    Basic static checks using regex and string rules.
    Returns a list of issues found.
    """
    issues = []

    # Check 1: Missing semicolons
    lines = java_code.split("\n")
    for i, line in enumerate(lines, start=1):
        if line.strip() and not line.strip().endswith(";") \
           and not line.strip().endswith("{") \
           and not line.strip().endswith("}") \
           and not line.strip().startswith("//") \
           and not line.strip().startswith("/*"):
            issues.append(f"Line {i}: Possible missing semicolon.")

    # Check 2: Naming conventions (class names must be PascalCase)
    class_names = re.findall(r'class\s+([A-Za-z0-9_]+)', java_code)
    for name in class_names:
        if not re.match(r'^[A-Z][A-Za-z0-9]+$', name):
            issues.append(f"Class name '{name}' does not follow PascalCase convention.")

    # Check 3: Method names (camelCase)
    method_names = re.findall(r'(?:public|private|protected)\s+[^\s]+\s+([a-zA-Z0-9_]+)\s*\(', java_code)
    for name in method_names:
        if not re.match(r'^[a-z][A-Za-z0-9]+$', name):
            issues.append(f"Method name '{name}' should follow camelCase convention.")

    # Check 4: Use of System.out.println (debugging practice)
    if "System.out.println" in java_code:
        issues.append("Avoid using System.out.println in production code. Use Logger instead.")

    # Check 5: Empty catch blocks
    if re.search(r'catch\s*\(.*\)\s*\{\s*\}', java_code):
        issues.append("Empty catch block found. Handle exceptions properly.")

    return issues

```

```python
from openai import OpenAI

client = OpenAI()

def ai_suggestions(java_code, static_issues):
    """
    Ask AI to provide improvements based on static issues.
    """
    prompt = f"""
    You are a senior Java reviewer.
    The following static analysis issues were detected:
    {static_issues}

    Based on these findings, suggest clear improvements or refactoring ideas.
    Provide code examples if needed.

    Code:
    {java_code}
    """

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "You are an expert in Java code review and static analysis."},
            {"role": "user", "content": prompt}
        ]
    )
    return response.choices[0].message.content
```

```python
if __name__ == "__main__":
    java_code = open("data/Calculator.java").read()

    print("\n🔍 Running static analysis...")
    static_issues = static_checks(java_code)
    print("Issues found:")
    for issue in static_issues:
        print(" -", issue)

    print("\n🤖 Generating AI suggestions...")
    suggestions = ai_suggestions(java_code, static_issues)

    print("\n=== AI Suggestions ===\n")
    print(suggestions)
```
